Window Based Approach

```python
print("Installing dependencies (this may take 2-3 minutes)...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION
# ============================================
WINDOW_SIZE = 2.0  # seconds - sliding window for speaker embedding
WINDOW_STEP = 0.5  # seconds - step size for sliding window
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s")
print(f"  Step size: {WINDOW_STEP}s")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 10 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix (too large to show all)
sample_size = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size, :sample_size],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: CLUSTERING WINDOWS TO IDENTIFY SPEAKERS")
print("="*60)

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔍 Auto-detecting number of speakers...")

    # Calculate eigenvalue gap
    from scipy.linalg import eigh
    eigenvalues, _ = eigh(similarity_matrix)
    eigenvalues = np.sort(eigenvalues)[::-1]

    max_speakers = min(8, len(X))
    eigengaps = []
    for i in range(1, min(max_speakers, len(eigenvalues)-1)):
        gap = eigenvalues[i-1] - eigenvalues[i]
        eigengaps.append(gap)

    best_n = np.argmax(eigengaps) + 1
    num_speakers = max(2, min(best_n, max_speakers))
    print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speakers...")
clustering = SpectralClustering(
    n_clusters=num_speakers,
    affinity='precomputed',
    assign_labels='kmeans',
    random_state=42
)
window_labels = clustering.fit_predict(similarity_matrix)

print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")

# For each word, find the closest window and assign its speaker
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find closest window center
    closest_idx = 0
    min_dist = float('inf')

    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        if dist < min_dist:
            min_dist = dist
            closest_idx = i

    word['speaker'] = window_labels[closest_idx]

print(f"✓ Speaker assignment complete")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```

```python

print("🔧 Fixing NumPy compatibility issue...")

# CRITICAL FIX: Install compatible NumPy version FIRST
!pip install -q numpy==2.0.2

print("Installing other dependencies...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION (OPTIMIZED FOR SHORT WORDS)
# ============================================
WINDOW_SIZE = 1.5   # seconds - REDUCED for better short-word detection
WINDOW_STEP = 0.25  # seconds - SMALLER step for finer temporal resolution
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds
VOTING_WINDOW = 3   # Number of nearby windows to consider for voting

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s (optimized for short words)")
print(f"  Step size: {WINDOW_STEP}s (finer temporal resolution)")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 20 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix
sample_size_plot = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size_plot, :sample_size_plot],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size_plot} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: CLUSTERING WINDOWS TO IDENTIFY SPEAKERS")
print("="*60)

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔍 Auto-detecting number of speakers...")

    # Calculate eigenvalue gap
    from scipy.linalg import eigh
    eigenvalues, _ = eigh(similarity_matrix)
    eigenvalues = np.sort(eigenvalues)[::-1]

    max_speakers = min(8, len(X))
    eigengaps = []
    for i in range(1, min(max_speakers, len(eigenvalues)-1)):
        gap = eigenvalues[i-1] - eigenvalues[i]
        eigengaps.append(gap)

    best_n = np.argmax(eigengaps) + 1
    num_speakers = max(2, min(best_n, max_speakers))
    print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speakers...")
clustering = SpectralClustering(
    n_clusters=num_speakers,
    affinity='precomputed',
    assign_labels='kmeans',
    random_state=42
)
window_labels = clustering.fit_predict(similarity_matrix)

print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS (WITH VOTING)
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")
print(f"  Using weighted voting from {VOTING_WINDOW} nearest windows")

# For each word, use VOTING from multiple nearby windows
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find distances to all windows
    distances = []
    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        distances.append((dist, i))

    # Sort by distance and take top VOTING_WINDOW closest windows
    distances.sort()
    closest_windows = distances[:min(VOTING_WINDOW, len(distances))]

    # Vote: count which speaker label appears most
    votes = {}
    for dist, idx in closest_windows:
        speaker = window_labels[idx]
        # Weight vote by inverse distance (closer = more weight)
        weight = 1.0 / (dist + 0.1)  # +0.1 to avoid division by zero
        votes[speaker] = votes.get(speaker, 0) + weight

    # Assign speaker with highest weighted vote
    word['speaker'] = max(votes, key=votes.get)

print(f"✓ Speaker assignment complete (using weighted voting)")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# ============================================
# POST-PROCESSING: FIX SHORT ISOLATED SEGMENTS
# ============================================
print(f"\n🔧 Post-processing: Fixing short isolated segments...")

# Fix segments shorter than 0.5s that are sandwiched between same speaker
fixed_count = 0
i = 1
while i < len(speaker_segments) - 1:
    current = speaker_segments[i]
    prev = speaker_segments[i-1]
    next_seg = speaker_segments[i+1]

    duration = current['end'] - current['start']

    # If current segment is short and surrounded by same speaker, merge it
    if duration < 0.5 and prev['speaker'] == next_seg['speaker'] and current['speaker'] != prev['speaker']:
        # Reassign to surrounding speaker
        current['speaker'] = prev['speaker']
        fixed_count += 1

    i += 1

# Re-merge consecutive segments with same speaker after fixing
final_segments = []
current_segment = None

for seg in speaker_segments:
    if current_segment is None or current_segment['speaker'] != seg['speaker']:
        if current_segment is not None:
            final_segments.append(current_segment)
        current_segment = {
            'speaker': seg['speaker'],
            'start': seg['start'],
            'end': seg['end'],
            'words': seg['words'].copy(),
            'text': seg['text']
        }
    else:
        # Merge with current
        current_segment['end'] = seg['end']
        current_segment['words'].extend(seg['words'])
        current_segment['text'] = ' '.join(current_segment['words']).strip()

if current_segment is not None:
    final_segments.append(current_segment)

speaker_segments = final_segments

print(f"✓ Fixed {fixed_count} short isolated segments")
print(f"✓ Final segment count: {len(speaker_segments)}")

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print(f"\n🎯 Improvements applied:")
print(f"  • Window size: {WINDOW_SIZE}s (smaller for short words)")
print(f"  • Step size: {WINDOW_STEP}s (finer resolution)")
print(f"  • Weighted voting from {VOTING_WINDOW} nearest windows")
print(f"  • Post-processing fixed {fixed_count} isolated segments")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```

```python
print("🔧 Fixing NumPy compatibility issue...")

# CRITICAL FIX: Install compatible NumPy version FIRST
!pip install -q numpy==2.0.2

print("Installing other dependencies...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION (OPTIMIZED FOR SHORT WORDS)
# ============================================
WINDOW_SIZE = 1.5   # seconds - REDUCED for better short-word detection
WINDOW_STEP = 0.25  # seconds - SMALLER step for finer temporal resolution
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds
VOTING_WINDOW = 3   # Number of nearby windows to consider for voting

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s (optimized for short words)")
print(f"  Step size: {WINDOW_STEP}s (finer temporal resolution)")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 20 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix
sample_size_plot = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size_plot, :sample_size_plot],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size_plot} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: IDENTIFYING SPEAKERS")
print("="*60)

# PRE-CHECK: Analyze embedding diversity to detect monologue
print("\n🔍 Analyzing audio characteristics...")

# Calculate pairwise distances between random sample of embeddings
sample_size = min(50, len(X))
sample_indices = np.random.choice(len(X), sample_size, replace=False)
sample_embeddings = X[sample_indices]

# Calculate all pairwise distances
pairwise_distances = cdist(sample_embeddings, sample_embeddings, metric='cosine')
# Remove diagonal (self-comparisons)
mask = np.ones(pairwise_distances.shape, dtype=bool)
np.fill_diagonal(mask, False)
pairwise_distances_clean = pairwise_distances[mask]

# Calculate statistics
mean_distance = np.mean(pairwise_distances_clean)
std_distance = np.std(pairwise_distances_clean)
max_distance = np.max(pairwise_distances_clean)

print(f"  Mean voice distance: {mean_distance:.3f}")
print(f"  Distance std dev: {std_distance:.3f}")
print(f"  Max distance: {max_distance:.3f}")

# Monologue detection heuristic:
# - Low mean distance (<0.15) = very similar voices = likely monologue
# - Low std (<0.08) = consistent similarity = likely monologue
is_likely_monologue = (mean_distance < 0.15 and std_distance < 0.08)

if is_likely_monologue:
    print(f"  → Audio characteristics suggest MONOLOGUE")
    print(f"     (voices are very similar throughout)")
else:
    print(f"  → Audio characteristics suggest MULTIPLE SPEAKERS")
    print(f"     (sufficient voice diversity detected)")

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔄 Auto-detecting number of speakers...")

    # If pre-check suggests monologue, start with that assumption
    if is_likely_monologue:
        print(f"  Pre-analysis suggests monologue")
        num_speakers = 1
        print(f"  → Set to {num_speakers} speaker")
    else:
        # Calculate eigenvalue gap for multi-speaker detection
        from scipy.linalg import eigh
        eigenvalues, _ = eigh(similarity_matrix)
        eigenvalues = np.sort(eigenvalues)[::-1]

        max_speakers = min(8, len(X))
        eigengaps = []
        for i in range(1, min(max_speakers, len(eigenvalues)-1)):
            gap = eigenvalues[i-1] - eigenvalues[i]
            eigengaps.append(gap)

        # Get initial estimate
        best_n = np.argmax(eigengaps) + 1
        best_n = max(2, min(best_n, max_speakers))

        # For 2 speakers: Double-check with cross-similarity
        if best_n == 2:
            test_clustering = SpectralClustering(
                n_clusters=2,
                affinity='precomputed',
                assign_labels='kmeans',
                random_state=42
            )
            test_labels = test_clustering.fit_predict(similarity_matrix)

            # Calculate average similarity BETWEEN the two detected speaker clusters
            cluster_0_indices = np.where(test_labels == 0)[0]
            cluster_1_indices = np.where(test_labels == 1)[0]

            # Sample windows from each cluster and check cross-similarity
            sample_size = min(20, len(cluster_0_indices), len(cluster_1_indices))
            cross_similarities = []

            for i in np.random.choice(cluster_0_indices, sample_size, replace=False):
                for j in np.random.choice(cluster_1_indices, sample_size, replace=False):
                    cross_similarities.append(similarity_matrix[i, j])

            avg_cross_similarity = np.mean(cross_similarities)

            print(f"  Eigenvalue method: {best_n} speakers")
            print(f"  Cross-cluster similarity: {avg_cross_similarity:.3f}")

            # If clusters are very similar (>0.70), likely same person
            if avg_cross_similarity > 0.70:
                print(f"  → Clusters too similar - adjusting to monologue")
                num_speakers = 1
            else:
                num_speakers = 2
                print(f"  → Confirmed {num_speakers} distinct speakers")
        else:
            num_speakers = best_n
            print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speaker(s)...")

if num_speakers == 1:
    # All windows belong to same speaker
    window_labels = np.zeros(len(windows), dtype=int)
    print(f"✓ Single speaker mode - all windows assigned to SPEAKER_00")
else:
    clustering = SpectralClustering(
        n_clusters=num_speakers,
        affinity='precomputed',
        assign_labels='kmeans',
        random_state=42
    )
    window_labels = clustering.fit_predict(similarity_matrix)
    print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS (WITH VOTING)
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")
print(f"  Using weighted voting from {VOTING_WINDOW} nearest windows")

# For each word, use VOTING from multiple nearby windows
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find distances to all windows
    distances = []
    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        distances.append((dist, i))

    # Sort by distance and take top VOTING_WINDOW closest windows
    distances.sort()
    closest_windows = distances[:min(VOTING_WINDOW, len(distances))]

    # Vote: count which speaker label appears most
    votes = {}
    for dist, idx in closest_windows:
        speaker = window_labels[idx]
        # Weight vote by inverse distance (closer = more weight)
        weight = 1.0 / (dist + 0.1)  # +0.1 to avoid division by zero
        votes[speaker] = votes.get(speaker, 0) + weight

    # Assign speaker with highest weighted vote
    word['speaker'] = max(votes, key=votes.get)

print(f"✓ Speaker assignment complete (using weighted voting)")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# ============================================
# POST-PROCESSING: FIX SHORT ISOLATED SEGMENTS
# ============================================
print(f"\n🔧 Post-processing: Fixing short isolated segments...")

# Fix segments shorter than 0.5s that are sandwiched between same speaker
fixed_count = 0
i = 1
while i < len(speaker_segments) - 1:
    current = speaker_segments[i]
    prev = speaker_segments[i-1]
    next_seg = speaker_segments[i+1]

    duration = current['end'] - current['start']

    # If current segment is short and surrounded by same speaker, merge it
    if duration < 0.5 and prev['speaker'] == next_seg['speaker'] and current['speaker'] != prev['speaker']:
        # Reassign to surrounding speaker
        current['speaker'] = prev['speaker']
        fixed_count += 1

    i += 1

# Re-merge consecutive segments with same speaker after fixing
final_segments = []
current_segment = None

for seg in speaker_segments:
    if current_segment is None or current_segment['speaker'] != seg['speaker']:
        if current_segment is not None:
            final_segments.append(current_segment)
        current_segment = {
            'speaker': seg['speaker'],
            'start': seg['start'],
            'end': seg['end'],
            'words': seg['words'].copy(),
            'text': seg['text']
        }
    else:
        # Merge with current
        current_segment['end'] = seg['end']
        current_segment['words'].extend(seg['words'])
        current_segment['text'] = ' '.join(current_segment['words']).strip()

if current_segment is not None:
    final_segments.append(current_segment)

speaker_segments = final_segments

print(f"✓ Fixed {fixed_count} short isolated segments")
print(f"✓ Final segment count: {len(speaker_segments)}")

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print(f"\n🎯 Improvements applied:")
print(f"  • Window size: {WINDOW_SIZE}s (smaller for short words)")
print(f"  • Step size: {WINDOW_STEP}s (finer resolution)")
print(f"  • Weighted voting from {VOTING_WINDOW} nearest windows")
print(f"  • Post-processing fixed {fixed_count} isolated segments")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```

```python
print("🔧 Fixing NumPy compatibility issue...")

# CRITICAL FIX: Install compatible NumPy version FIRST
!pip install -q numpy==2.0.2

print("Installing other dependencies...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION (OPTIMIZED FOR SHORT WORDS)
# ============================================
WINDOW_SIZE = 1.5   # seconds - REDUCED for better short-word detection
WINDOW_STEP = 0.25  # seconds - SMALLER step for finer temporal resolution
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds
VOTING_WINDOW = 3   # Number of nearby windows to consider for voting

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s (optimized for short words)")
print(f"  Step size: {WINDOW_STEP}s (finer temporal resolution)")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 20 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix
sample_size_plot = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size_plot, :sample_size_plot],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size_plot} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: IDENTIFYING SPEAKERS")
print("="*60)

# PRE-CHECK: Analyze embedding diversity to detect monologue
print("\n🔍 Analyzing audio characteristics...")

# Calculate pairwise distances between random sample of embeddings
sample_size = min(50, len(X))
sample_indices = np.random.choice(len(X), sample_size, replace=False)
sample_embeddings = X[sample_indices]

# Calculate all pairwise distances
pairwise_distances = cdist(sample_embeddings, sample_embeddings, metric='cosine')
# Remove diagonal (self-comparisons)
mask = np.ones(pairwise_distances.shape, dtype=bool)
np.fill_diagonal(mask, False)
pairwise_distances_clean = pairwise_distances[mask]

# Calculate statistics
mean_distance = np.mean(pairwise_distances_clean)
std_distance = np.std(pairwise_distances_clean)
max_distance = np.max(pairwise_distances_clean)

print(f"  Mean voice distance: {mean_distance:.3f}")
print(f"  Distance std dev: {std_distance:.3f}")
print(f"  Max distance: {max_distance:.3f}")

# Monologue detection heuristic:
# - Low mean distance (<0.15) = very similar voices = likely monologue
# - Low std (<0.08) = consistent similarity = likely monologue
is_likely_monologue = (mean_distance < 0.15 and std_distance < 0.08)

if is_likely_monologue:
    print(f"  → Audio characteristics suggest MONOLOGUE")
    print(f"     (voices are very similar throughout)")
else:
    print(f"  → Audio characteristics suggest MULTIPLE SPEAKERS")
    print(f"     (sufficient voice diversity detected)")

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔄 Auto-detecting number of speakers...")

    # If pre-check suggests monologue, start with that assumption
    if is_likely_monologue:
        print(f"  Pre-analysis suggests monologue")
        num_speakers = 1
        print(f"  → Set to {num_speakers} speaker")
    else:
        # Calculate eigenvalue gap for multi-speaker detection
        from scipy.linalg import eigh
        eigenvalues, _ = eigh(similarity_matrix)
        eigenvalues = np.sort(eigenvalues)[::-1]

        max_speakers = min(8, len(X))
        eigengaps = []
        for i in range(1, min(max_speakers, len(eigenvalues)-1)):
            gap = eigenvalues[i-1] - eigenvalues[i]
            eigengaps.append(gap)

        # Get initial estimate
        best_n = np.argmax(eigengaps) + 1
        best_n = max(2, min(best_n, max_speakers))

        # For 2 speakers: Double-check with cross-similarity
        if best_n == 2:
            test_clustering = SpectralClustering(
                n_clusters=2,
                affinity='precomputed',
                assign_labels='kmeans',
                random_state=42
            )
            test_labels = test_clustering.fit_predict(similarity_matrix)

            # Calculate average similarity BETWEEN the two detected speaker clusters
            cluster_0_indices = np.where(test_labels == 0)[0]
            cluster_1_indices = np.where(test_labels == 1)[0]

            # Sample windows from each cluster and check cross-similarity
            sample_size = min(20, len(cluster_0_indices), len(cluster_1_indices))
            cross_similarities = []

            for i in np.random.choice(cluster_0_indices, sample_size, replace=False):
                for j in np.random.choice(cluster_1_indices, sample_size, replace=False):
                    cross_similarities.append(similarity_matrix[i, j])

            avg_cross_similarity = np.mean(cross_similarities)

            print(f"  Eigenvalue method: {best_n} speakers")
            print(f"  Cross-cluster similarity: {avg_cross_similarity:.3f}")

            # If clusters are very similar (>0.70), likely same person
            if avg_cross_similarity > 0.70:
                print(f"  → Clusters too similar - adjusting to monologue")
                num_speakers = 1
            else:
                num_speakers = 2
                print(f"  → Confirmed {num_speakers} distinct speakers")
        else:
            num_speakers = best_n
            print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speaker(s)...")

if num_speakers == 1:
    # All windows belong to same speaker
    window_labels = np.zeros(len(windows), dtype=int)
    print(f"✓ Single speaker mode - all windows assigned to SPEAKER_00")
else:
    clustering = SpectralClustering(
        n_clusters=num_speakers,
        affinity='precomputed',
        assign_labels='kmeans',
        random_state=42
    )
    window_labels = clustering.fit_predict(similarity_matrix)
    print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS (WITH VOTING)
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")
print(f"  Using weighted voting from {VOTING_WINDOW} nearest windows")

# For each word, use VOTING from multiple nearby windows
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find distances to all windows
    distances = []
    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        distances.append((dist, i))

    # Sort by distance and take top VOTING_WINDOW closest windows
    distances.sort()
    closest_windows = distances[:min(VOTING_WINDOW, len(distances))]

    # Vote: count which speaker label appears most
    votes = {}
    for dist, idx in closest_windows:
        speaker = window_labels[idx]
        # Weight vote by inverse distance (closer = more weight)
        weight = 1.0 / (dist + 0.1)  # +0.1 to avoid division by zero
        votes[speaker] = votes.get(speaker, 0) + weight

    # Assign speaker with highest weighted vote
    word['speaker'] = max(votes, key=votes.get)

print(f"✓ Speaker assignment complete (using weighted voting)")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# ============================================
# POST-PROCESSING: FIX SHORT ISOLATED SEGMENTS
# ============================================
print(f"\n🔧 Post-processing: Fixing short isolated segments...")

# Fix segments shorter than 0.5s that are sandwiched between same speaker
fixed_count = 0
i = 1
while i < len(speaker_segments) - 1:
    current = speaker_segments[i]
    prev = speaker_segments[i-1]
    next_seg = speaker_segments[i+1]

    duration = current['end'] - current['start']

    # If current segment is short and surrounded by same speaker, merge it
    if duration < 0.5 and prev['speaker'] == next_seg['speaker'] and current['speaker'] != prev['speaker']:
        # Reassign to surrounding speaker
        current['speaker'] = prev['speaker']
        fixed_count += 1

    i += 1

# Re-merge consecutive segments with same speaker after fixing
final_segments = []
current_segment = None

for seg in speaker_segments:
    if current_segment is None or current_segment['speaker'] != seg['speaker']:
        if current_segment is not None:
            final_segments.append(current_segment)
        current_segment = {
            'speaker': seg['speaker'],
            'start': seg['start'],
            'end': seg['end'],
            'words': seg['words'].copy(),
            'text': seg['text']
        }
    else:
        # Merge with current
        current_segment['end'] = seg['end']
        current_segment['words'].extend(seg['words'])
        current_segment['text'] = ' '.join(current_segment['words']).strip()

if current_segment is not None:
    final_segments.append(current_segment)

speaker_segments = final_segments

print(f"✓ Fixed {fixed_count} short isolated segments")
print(f"✓ Final segment count: {len(speaker_segments)}")

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print(f"\n🎯 Improvements applied:")
print(f"  • Window size: {WINDOW_SIZE}s (smaller for short words)")
print(f"  • Step size: {WINDOW_STEP}s (finer resolution)")
print(f"  • Weighted voting from {VOTING_WINDOW} nearest windows")
print(f"  • Post-processing fixed {fixed_count} isolated segments")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```

```python
print("🔧 Fixing NumPy compatibility issue...")

# CRITICAL FIX: Install compatible NumPy version FIRST
!pip install -q numpy==2.0.2

print("Installing other dependencies...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION (OPTIMIZED FOR SHORT WORDS)
# ============================================
WINDOW_SIZE = 1.5   # seconds - REDUCED for better short-word detection
WINDOW_STEP = 0.25  # seconds - SMALLER step for finer temporal resolution
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds
VOTING_WINDOW = 3   # Number of nearby windows to consider for voting

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s (optimized for short words)")
print(f"  Step size: {WINDOW_STEP}s (finer temporal resolution)")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 20 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix
sample_size_plot = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size_plot, :sample_size_plot],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size_plot} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: IDENTIFYING SPEAKERS")
print("="*60)

# PRE-CHECK: Analyze embedding diversity to detect monologue
print("\n🔍 Analyzing audio characteristics...")

# Calculate pairwise distances between random sample of embeddings
sample_size = min(50, len(X))
sample_indices = np.random.choice(len(X), sample_size, replace=False)
sample_embeddings = X[sample_indices]

# Calculate all pairwise distances
pairwise_distances = cdist(sample_embeddings, sample_embeddings, metric='cosine')
# Remove diagonal (self-comparisons)
mask = np.ones(pairwise_distances.shape, dtype=bool)
np.fill_diagonal(mask, False)
pairwise_distances_clean = pairwise_distances[mask]

# Calculate statistics
mean_distance = np.mean(pairwise_distances_clean)
std_distance = np.std(pairwise_distances_clean)
max_distance = np.max(pairwise_distances_clean)

print(f"  Mean voice distance: {mean_distance:.3f}")
print(f"  Distance std dev: {std_distance:.3f}")
print(f"  Max distance: {max_distance:.3f}")

# Monologue detection heuristic:
# - Low mean distance (<0.15) = very similar voices = likely monologue
# - Low std (<0.08) = consistent similarity = likely monologue
is_likely_monologue = (mean_distance < 0.15 and std_distance < 0.08)

if is_likely_monologue:
    print(f"  → Audio characteristics suggest MONOLOGUE")
    print(f"     (voices are very similar throughout)")
else:
    print(f"  → Audio characteristics suggest MULTIPLE SPEAKERS")
    print(f"     (sufficient voice diversity detected)")

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔄 Auto-detecting number of speakers...")

    # If pre-check suggests monologue, start with that assumption
    if is_likely_monologue:
        print(f"  Pre-analysis suggests monologue")
        num_speakers = 1
        print(f"  → Set to {num_speakers} speaker")
    else:
        # Calculate eigenvalue gap for multi-speaker detection
        from scipy.linalg import eigh
        eigenvalues, _ = eigh(similarity_matrix)
        eigenvalues = np.sort(eigenvalues)[::-1]

        max_speakers = min(8, len(X))
        eigengaps = []
        for i in range(1, min(max_speakers, len(eigenvalues)-1)):
            gap = eigenvalues[i-1] - eigenvalues[i]
            eigengaps.append(gap)

        # Get initial estimate
        best_n = np.argmax(eigengaps) + 1
        best_n = max(2, min(best_n, max_speakers))

        # For 2 speakers: Double-check with cross-similarity
        if best_n == 2:
            test_clustering = SpectralClustering(
                n_clusters=2,
                affinity='precomputed',
                assign_labels='kmeans',
                random_state=42
            )
            test_labels = test_clustering.fit_predict(similarity_matrix)

            # Calculate average similarity BETWEEN the two detected speaker clusters
            cluster_0_indices = np.where(test_labels == 0)[0]
            cluster_1_indices = np.where(test_labels == 1)[0]

            # Sample windows from each cluster and check cross-similarity
            sample_size = min(20, len(cluster_0_indices), len(cluster_1_indices))
            cross_similarities = []

            for i in np.random.choice(cluster_0_indices, sample_size, replace=False):
                for j in np.random.choice(cluster_1_indices, sample_size, replace=False):
                    cross_similarities.append(similarity_matrix[i, j])

            avg_cross_similarity = np.mean(cross_similarities)

            print(f"  Eigenvalue method: {best_n} speakers")
            print(f"  Cross-cluster similarity: {avg_cross_similarity:.3f}")

            # If clusters are very similar (>0.70), likely same person
            if avg_cross_similarity > 0.70:
                print(f"  → Clusters too similar - adjusting to monologue")
                num_speakers = 1
            else:
                num_speakers = 2
                print(f"  → Confirmed {num_speakers} distinct speakers")
        else:
            num_speakers = best_n
            print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speaker(s)...")

if num_speakers == 1:
    # All windows belong to same speaker
    window_labels = np.zeros(len(windows), dtype=int)
    print(f"✓ Single speaker mode - all windows assigned to SPEAKER_00")
else:
    clustering = SpectralClustering(
        n_clusters=num_speakers,
        affinity='precomputed',
        assign_labels='kmeans',
        random_state=42
    )
    window_labels = clustering.fit_predict(similarity_matrix)
    print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS (WITH VOTING)
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")
print(f"  Using weighted voting from {VOTING_WINDOW} nearest windows")

# For each word, use VOTING from multiple nearby windows
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find distances to all windows
    distances = []
    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        distances.append((dist, i))

    # Sort by distance and take top VOTING_WINDOW closest windows
    distances.sort()
    closest_windows = distances[:min(VOTING_WINDOW, len(distances))]

    # Vote: count which speaker label appears most
    votes = {}
    for dist, idx in closest_windows:
        speaker = window_labels[idx]
        # Weight vote by inverse distance (closer = more weight)
        weight = 1.0 / (dist + 0.1)  # +0.1 to avoid division by zero
        votes[speaker] = votes.get(speaker, 0) + weight

    # Assign speaker with highest weighted vote
    word['speaker'] = max(votes, key=votes.get)

print(f"✓ Speaker assignment complete (using weighted voting)")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# ============================================
# POST-PROCESSING: FIX SHORT ISOLATED SEGMENTS
# ============================================
print(f"\n🔧 Post-processing: Fixing short isolated segments...")

# Fix segments shorter than 0.5s that are sandwiched between same speaker
fixed_count = 0
i = 1
while i < len(speaker_segments) - 1:
    current = speaker_segments[i]
    prev = speaker_segments[i-1]
    next_seg = speaker_segments[i+1]

    duration = current['end'] - current['start']

    # If current segment is short and surrounded by same speaker, merge it
    if duration < 0.5 and prev['speaker'] == next_seg['speaker'] and current['speaker'] != prev['speaker']:
        # Reassign to surrounding speaker
        current['speaker'] = prev['speaker']
        fixed_count += 1

    i += 1

# Re-merge consecutive segments with same speaker after fixing
final_segments = []
current_segment = None

for seg in speaker_segments:
    if current_segment is None or current_segment['speaker'] != seg['speaker']:
        if current_segment is not None:
            final_segments.append(current_segment)
        current_segment = {
            'speaker': seg['speaker'],
            'start': seg['start'],
            'end': seg['end'],
            'words': seg['words'].copy(),
            'text': seg['text']
        }
    else:
        # Merge with current
        current_segment['end'] = seg['end']
        current_segment['words'].extend(seg['words'])
        current_segment['text'] = ' '.join(current_segment['words']).strip()

if current_segment is not None:
    final_segments.append(current_segment)

speaker_segments = final_segments

print(f"✓ Fixed {fixed_count} short isolated segments")
print(f"✓ Final segment count: {len(speaker_segments)}")

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print(f"\n🎯 Improvements applied:")
print(f"  • Window size: {WINDOW_SIZE}s (smaller for short words)")
print(f"  • Step size: {WINDOW_STEP}s (finer resolution)")
print(f"  • Weighted voting from {VOTING_WINDOW} nearest windows")
print(f"  • Post-processing fixed {fixed_count} isolated segments")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```

```python
print("🔧 Fixing NumPy compatibility issue...")

# CRITICAL FIX: Install compatible NumPy version FIRST
!pip install -q numpy==2.0.2

print("Installing other dependencies...")

!pip install -q openai-whisper
!pip install -q resemblyzer
!pip install -q scikit-learn
!pip install -q librosa
!pip install -q pydub
!pip install -q matplotlib
!pip install -q seaborn

import whisper
import torch
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav
from sklearn.cluster import SpectralClustering
from scipy.spatial.distance import cosine, cdist
import librosa
import warnings
import json
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

print(f"\n✓ Installation complete!")
print(f"GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device: {torch.cuda.get_device_name(0)}")

# ============================================
# CONFIGURATION (OPTIMIZED FOR SHORT WORDS)
# ============================================
WINDOW_SIZE = 1.5   # seconds - REDUCED for better short-word detection
WINDOW_STEP = 0.25  # seconds - SMALLER step for finer temporal resolution
MIN_SEGMENT_LENGTH = 0.5  # minimum segment length in seconds
VOTING_WINDOW = 3   # Number of nearby windows to consider for voting

# ============================================
# STEP 2: UPLOAD AUDIO FILE
# ============================================
from google.colab import files

print("\n" + "="*60)
print("UPLOAD YOUR AUDIO FILE")
print("="*60)
uploaded = files.upload()

if not uploaded:
    raise Exception("No file uploaded!")

audio_file = list(uploaded.keys())[0]
print(f"✓ Uploaded: {audio_file}")
print(f"  File size: {len(uploaded[audio_file]) / 1024:.2f} KB")

# ============================================
# STEP 3: TRANSCRIBE WITH WHISPER
# ============================================
print("\n" + "="*60)
print("STEP 1: TRANSCRIBING WITH WHISPER")
print("="*60)

audio_data_preview, sr_preview = librosa.load(audio_file, sr=16000)
duration = len(audio_data_preview) / sr_preview
print(f"\n📊 Audio Information:")
print(f"  Duration: {duration:.2f} seconds")
print(f"  Sample Rate: {sr_preview} Hz")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"\n🔄 Loading Whisper model on {device}...")
model = whisper.load_model("base", device=device)

print("🎤 Transcribing audio with word-level timestamps...")
result = model.transcribe(audio_file, word_timestamps=True)

print(f"\n✓ Transcription complete!")
print(f"  Language: {result['language']}")
print(f"  Total segments: {len(result['segments'])}")

# Extract all words with timestamps
all_words = []
for seg in result['segments']:
    if 'words' in seg:
        for word in seg['words']:
            all_words.append(word)

print(f"  Total words: {len(all_words)}")

# Show sample word timestamps
print(f"\n📝 Sample Word Timestamps (first 10 words):")
for i, word in enumerate(all_words[:10]):
    print(f"  [{word['start']:.1f}s–{word['end']:.1f}s] {word['word']}")

# ============================================
# STEP 4: SLIDING WINDOW SPEAKER EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("STEP 2: EXTRACTING SPEAKER EMBEDDINGS (SLIDING WINDOW)")
print("="*60)

audio_data, sr = librosa.load(audio_file, sr=16000)
print(f"\n🔄 Loading voice encoder model...")
encoder = VoiceEncoder()
print(f"✓ Voice encoder loaded")

# Create sliding windows across the entire audio
windows = []
embeddings = []

print(f"\n🎯 Creating sliding windows:")
print(f"  Window size: {WINDOW_SIZE}s (optimized for short words)")
print(f"  Step size: {WINDOW_STEP}s (finer temporal resolution)")

current_time = 0
window_count = 0

while current_time + WINDOW_SIZE <= duration:
    start_time = current_time
    end_time = current_time + WINDOW_SIZE

    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)
    audio_chunk = audio_data[start_sample:end_sample]

    try:
        processed = preprocess_wav(audio_chunk, sr)
        emb = encoder.embed_utterance(processed)
        embeddings.append(emb)
        windows.append({'start': start_time, 'end': end_time, 'center': (start_time + end_time) / 2})
        window_count += 1

        if window_count % 20 == 0:
            print(f"  Processed {window_count} windows...")
    except:
        pass

    current_time += WINDOW_STEP

print(f"\n✓ Extracted {len(embeddings)} window embeddings")

# Show sample embeddings
print(f"\n📊 Sample Embeddings (first 3 windows):")
for i in range(min(3, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s]:")
    print(f"    Embedding vector (first 5 dims): {embeddings[i][:5]}")
    print(f"    Embedding norm: {np.linalg.norm(embeddings[i]):.4f}")

# ============================================
# VISUALIZE EMBEDDINGS
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: EMBEDDING SIMILARITY")
print("="*60)

X = np.array(embeddings)

# Compute similarity matrix
similarity_matrix = 1 - cdist(X, X, metric='cosine')
print(f"\n📊 Similarity matrix computed: {similarity_matrix.shape}")

# Show sample similarity values
print(f"\n📊 Sample Similarity Matrix (first 5x5):")
sample_size = min(5, len(X))
print("     ", end="")
for i in range(sample_size):
    print(f"W{i+1:2d}   ", end="")
print()
for i in range(sample_size):
    print(f"W{i+1:2d} ", end="")
    for j in range(sample_size):
        print(f"{similarity_matrix[i,j]:.2f}  ", end="")
    print()

print(f"\n💡 Interpretation:")
print(f"  High values (>0.80) = likely same speaker")
print(f"  Low values (<0.70) = likely different speakers")

# Show a sample of the similarity matrix
sample_size_plot = min(20, len(X))
plt.figure(figsize=(10, 8))
sns.heatmap(similarity_matrix[:sample_size_plot, :sample_size_plot],
            annot=True,
            fmt='.2f',
            cmap='RdYlGn',
            vmin=0, vmax=1,
            cbar_kws={'label': 'Cosine Similarity'})
plt.title(f'Speaker Embedding Similarity (First {sample_size_plot} windows)\n(Green = Similar, Red = Different)')
plt.xlabel('Window Index')
plt.ylabel('Window Index')
plt.tight_layout()
plt.show()

# ============================================
# STEP 5: CLUSTER WINDOWS TO IDENTIFY SPEAKERS
# ============================================
print("\n" + "="*60)
print("STEP 3: IDENTIFYING SPEAKERS")
print("="*60)

# PRE-CHECK: Analyze embedding diversity to detect monologue
print("\n🔍 Analyzing audio characteristics...")

# Calculate pairwise distances between random sample of embeddings
sample_size = min(50, len(X))
sample_indices = np.random.choice(len(X), sample_size, replace=False)
sample_embeddings = X[sample_indices]

# Calculate all pairwise distances
pairwise_distances = cdist(sample_embeddings, sample_embeddings, metric='cosine')
# Remove diagonal (self-comparisons)
mask = np.ones(pairwise_distances.shape, dtype=bool)
np.fill_diagonal(mask, False)
pairwise_distances_clean = pairwise_distances[mask]

# Calculate statistics
mean_distance = np.mean(pairwise_distances_clean)
std_distance = np.std(pairwise_distances_clean)
max_distance = np.max(pairwise_distances_clean)

print(f"  Mean voice distance: {mean_distance:.3f}")
print(f"  Distance std dev: {std_distance:.3f}")
print(f"  Max distance: {max_distance:.3f}")

# Monologue detection heuristic:
# - Low mean distance (<0.15) = very similar voices = likely monologue
# - Low std (<0.08) = consistent similarity = likely monologue
is_likely_monologue = (mean_distance < 0.15 and std_distance < 0.08)

if is_likely_monologue:
    print(f"  → Audio characteristics suggest MONOLOGUE")
    print(f"     (voices are very similar throughout)")
else:
    print(f"  → Audio characteristics suggest MULTIPLE SPEAKERS")
    print(f"     (sufficient voice diversity detected)")

num_speakers_input = input("\nHow many speakers? (press Enter for auto-detect): ").strip()

if num_speakers_input:
    num_speakers = int(num_speakers_input)
    print(f"\n✓ Using user-specified: {num_speakers} speakers")
else:
    print("\n🔄 Auto-detecting number of speakers...")

    # If pre-check suggests monologue, start with that assumption
    if is_likely_monologue:
        print(f"  Pre-analysis suggests monologue")
        num_speakers = 1
        print(f"  → Set to {num_speakers} speaker")
    else:
        # Calculate eigenvalue gap for multi-speaker detection
        from scipy.linalg import eigh
        eigenvalues, _ = eigh(similarity_matrix)
        eigenvalues = np.sort(eigenvalues)[::-1]

        max_speakers = min(8, len(X))
        eigengaps = []
        for i in range(1, min(max_speakers, len(eigenvalues)-1)):
            gap = eigenvalues[i-1] - eigenvalues[i]
            eigengaps.append(gap)

        # Get initial estimate
        best_n = np.argmax(eigengaps) + 1
        best_n = max(2, min(best_n, max_speakers))

        # For 2 speakers: Double-check with cross-similarity
        if best_n == 2:
            test_clustering = SpectralClustering(
                n_clusters=2,
                affinity='precomputed',
                assign_labels='kmeans',
                random_state=42
            )
            test_labels = test_clustering.fit_predict(similarity_matrix)

            # Calculate average similarity BETWEEN the two detected speaker clusters
            cluster_0_indices = np.where(test_labels == 0)[0]
            cluster_1_indices = np.where(test_labels == 1)[0]

            # Sample windows from each cluster and check cross-similarity
            sample_size = min(20, len(cluster_0_indices), len(cluster_1_indices))
            cross_similarities = []

            for i in np.random.choice(cluster_0_indices, sample_size, replace=False):
                for j in np.random.choice(cluster_1_indices, sample_size, replace=False):
                    cross_similarities.append(similarity_matrix[i, j])

            avg_cross_similarity = np.mean(cross_similarities)

            print(f"  Eigenvalue method: {best_n} speakers")
            print(f"  Cross-cluster similarity: {avg_cross_similarity:.3f}")

            # If clusters are very similar (>0.70), likely same person
            if avg_cross_similarity > 0.70:
                print(f"  → Clusters too similar - adjusting to monologue")
                num_speakers = 1
            else:
                num_speakers = 2
                print(f"  → Confirmed {num_speakers} distinct speakers")
        else:
            num_speakers = best_n
            print(f"  Detected {num_speakers} speakers")

# Cluster windows
print(f"\n🔄 Clustering {len(windows)} windows into {num_speakers} speaker(s)...")

if num_speakers == 1:
    # All windows belong to same speaker
    window_labels = np.zeros(len(windows), dtype=int)
    print(f"✓ Single speaker mode - all windows assigned to SPEAKER_00")
else:
    clustering = SpectralClustering(
        n_clusters=num_speakers,
        affinity='precomputed',
        assign_labels='kmeans',
        random_state=42
    )
    window_labels = clustering.fit_predict(similarity_matrix)
    print(f"✓ Clustering complete")

# Show sample window assignments
print(f"\n📋 Sample Window-to-Speaker Assignments (first 10 windows):")
for i in range(min(10, len(windows))):
    print(f"  Window {i+1} [{windows[i]['start']:.1f}s–{windows[i]['end']:.1f}s] → Cluster {window_labels[i]} (SPEAKER_{window_labels[i]:02d})")

print(f"\n👥 Speaker Distribution:")
for i in range(num_speakers):
    count = np.sum(window_labels == i)
    print(f"  SPEAKER_{i:02d}: {count} windows ({count/len(window_labels)*100:.1f}%)")

# ============================================
# STEP 6: ASSIGN SPEAKERS TO WORDS (WITH VOTING)
# ============================================
print("\n" + "="*60)
print("STEP 4: ASSIGNING SPEAKERS TO WORDS")
print("="*60)

print(f"\n🔄 Mapping {len(all_words)} words to speaker labels...")
print(f"  Using weighted voting from {VOTING_WINDOW} nearest windows")

# For each word, use VOTING from multiple nearby windows
for word in all_words:
    word_time = (word['start'] + word['end']) / 2

    # Find distances to all windows
    distances = []
    for i, window in enumerate(windows):
        dist = abs(window['center'] - word_time)
        distances.append((dist, i))

    # Sort by distance and take top VOTING_WINDOW closest windows
    distances.sort()
    closest_windows = distances[:min(VOTING_WINDOW, len(distances))]

    # Vote: count which speaker label appears most
    votes = {}
    for dist, idx in closest_windows:
        speaker = window_labels[idx]
        # Weight vote by inverse distance (closer = more weight)
        weight = 1.0 / (dist + 0.1)  # +0.1 to avoid division by zero
        votes[speaker] = votes.get(speaker, 0) + weight

    # Assign speaker with highest weighted vote
    word['speaker'] = max(votes, key=votes.get)

print(f"✓ Speaker assignment complete (using weighted voting)")

# Show sample word-to-speaker assignments
print(f"\n📋 Sample Word-to-Speaker Assignments (first 15 words):")
for i in range(min(15, len(all_words))):
    word = all_words[i]
    print(f"  {word['word']:15s} → SPEAKER_{word['speaker']:02d}")

# ============================================
# STEP 7: GROUP WORDS INTO SPEAKER SEGMENTS
# ============================================
print("\n" + "="*60)
print("STEP 5: CREATING SPEAKER SEGMENTS")
print("="*60)

print(f"\n🔄 Grouping consecutive words by same speaker...")

# Group consecutive words with same speaker
speaker_segments = []
current_segment = None

for word in all_words:
    speaker = word['speaker']

    if current_segment is None or current_segment['speaker'] != speaker:
        # Start new segment
        if current_segment is not None:
            speaker_segments.append(current_segment)

        current_segment = {
            'speaker': f"SPEAKER_{speaker:02d}",
            'start': word['start'],
            'end': word['end'],
            'words': [word['word']]
        }
    else:
        # Continue current segment
        current_segment['end'] = word['end']
        current_segment['words'].append(word['word'])

# Add last segment
if current_segment is not None:
    speaker_segments.append(current_segment)

print(f"✓ Created {len(speaker_segments)} speaker segments")

# Combine text for each segment
for seg in speaker_segments:
    seg['text'] = ' '.join(seg['words']).strip()

# ============================================
# POST-PROCESSING: FIX SHORT ISOLATED SEGMENTS
# ============================================
print(f"\n🔧 Post-processing: Fixing short isolated segments...")

# Fix segments shorter than 0.5s that are sandwiched between same speaker
fixed_count = 0
i = 1
while i < len(speaker_segments) - 1:
    current = speaker_segments[i]
    prev = speaker_segments[i-1]
    next_seg = speaker_segments[i+1]

    duration = current['end'] - current['start']

    # If current segment is short and surrounded by same speaker, merge it
    if duration < 0.5 and prev['speaker'] == next_seg['speaker'] and current['speaker'] != prev['speaker']:
        # Reassign to surrounding speaker
        current['speaker'] = prev['speaker']
        fixed_count += 1

    i += 1

# Re-merge consecutive segments with same speaker after fixing
final_segments = []
current_segment = None

for seg in speaker_segments:
    if current_segment is None or current_segment['speaker'] != seg['speaker']:
        if current_segment is not None:
            final_segments.append(current_segment)
        current_segment = {
            'speaker': seg['speaker'],
            'start': seg['start'],
            'end': seg['end'],
            'words': seg['words'].copy(),
            'text': seg['text']
        }
    else:
        # Merge with current
        current_segment['end'] = seg['end']
        current_segment['words'].extend(seg['words'])
        current_segment['text'] = ' '.join(current_segment['words']).strip()

if current_segment is not None:
    final_segments.append(current_segment)

speaker_segments = final_segments

print(f"✓ Fixed {fixed_count} short isolated segments")
print(f"✓ Final segment count: {len(speaker_segments)}")

# Show sample segments
print(f"\n📋 Sample Speaker Segments (first 5):")
for i in range(min(5, len(speaker_segments))):
    seg = speaker_segments[i]
    print(f"  [{seg['speaker']}] [{seg['start']:.1f}s–{seg['end']:.1f}s] {seg['text'][:60]}{'...' if len(seg['text']) > 60 else ''}")

# ============================================
# VISUALIZATION: SPEAKER TIMELINE
# ============================================
print("\n" + "="*60)
print("VISUALIZATION: SPEAKER TIMELINE")
print("="*60)

plt.figure(figsize=(14, 4))
colors = plt.cm.Set3(np.linspace(0, 1, num_speakers))

for seg in speaker_segments:
    speaker_id = int(seg['speaker'].split('_')[1])
    plt.barh(0, seg['end']-seg['start'], left=seg['start'],
             height=0.5, color=colors[speaker_id],
             edgecolor='black', linewidth=0.5)

plt.yticks([])
plt.xlabel('Time (seconds)', fontsize=12)
plt.title('Speaker Diarization Timeline', fontsize=14, weight='bold')
plt.grid(True, alpha=0.3, axis='x')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=colors[i], label=f'SPEAKER_{i:02d}')
                   for i in range(num_speakers)]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# ============================================
# STEP 8: DISPLAY RESULTS
# ============================================
print("\n" + "="*60)
print("FINAL TRANSCRIPTION WITH SPEAKERS")
print("="*60 + "\n")

transcript_lines = []
current_speaker = None

for seg in speaker_segments:
    speaker = seg['speaker']
    start = seg['start']
    text = seg['text']

    if speaker != current_speaker:
        speaker_line = f"\n[{speaker}]"
        print(speaker_line)
        transcript_lines.append(speaker_line)
        current_speaker = speaker

    line = f"[{start:.1f}s] {text}"
    print(line)
    transcript_lines.append(line)

# ============================================
# STATISTICS
# ============================================
print("\n" + "="*60)
print("STATISTICS")
print("="*60)

for i in range(num_speakers):
    speaker_name = f"SPEAKER_{i:02d}"
    speaker_segs = [s for s in speaker_segments if s['speaker'] == speaker_name]
    total_time = sum([s['end'] - s['start'] for s in speaker_segs])
    total_words = sum([len(s['words']) for s in speaker_segs])

    print(f"\n{speaker_name}:")
    print(f"  Segments: {len(speaker_segs)}")
    print(f"  Total speaking time: {total_time:.1f}s ({total_time/duration*100:.1f}%)")
    print(f"  Total words: {total_words}")

# ============================================
# STEP 9: SAVE AND DOWNLOAD
# ============================================
print("\n" + "="*60)
print("SAVING RESULTS")
print("="*60)

# Save as TXT
output_file = "transcript_with_speakers.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(transcript_lines))
files.download(output_file)
print(f"✓ Downloaded: {output_file}")

# Save as JSON
json_output = "transcript_detailed.json"
with open(json_output, "w", encoding="utf-8") as f:
    json.dump({
        'language': result['language'],
        'segments': speaker_segments,
        'num_speakers': num_speakers
    }, f, indent=2, ensure_ascii=False)
files.download(json_output)
print(f"✓ Downloaded: {json_output}")

# Save as SRT
srt_output = "transcript.srt"
with open(srt_output, "w", encoding="utf-8") as f:
    for i, seg in enumerate(speaker_segments, 1):
        speaker = seg['speaker']
        start = seg['start']
        end = seg['end']
        text = seg['text']

        start_time = f"{int(start//3600):02d}:{int((start%3600)//60):02d}:{int(start%60):02d},{int((start%1)*1000):03d}"
        end_time = f"{int(end//3600):02d}:{int((end%3600)//60):02d}:{int(end%60):02d},{int((end%1)*1000):03d}"

        f.write(f"{i}\n{start_time} --> {end_time}\n[{speaker}] {text}\n\n")
files.download(srt_output)
print(f"✓ Downloaded: {srt_output}")

print("\n" + "="*60)
print("✅ ALL DONE!")
print("="*60)
print(f"\nProcessed {len(speaker_segments)} segments")
print(f"Identified {num_speakers} unique speakers")
print(f"\n🎯 Improvements applied:")
print(f"  • Window size: {WINDOW_SIZE}s (smaller for short words)")
print(f"  • Step size: {WINDOW_STEP}s (finer resolution)")
print(f"  • Weighted voting from {VOTING_WINDOW} nearest windows")
print(f"  • Post-processing fixed {fixed_count} isolated segments")
print("\nFiles downloaded:")
print("  • transcript_with_speakers.txt")
print("  • transcript_detailed.json")
print("  • transcript.srt")
```
